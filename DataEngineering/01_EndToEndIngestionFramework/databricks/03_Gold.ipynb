```python
from pyspark.sql.functions import col

# 1. Define Paths and Table Names
# We point to the 'gold' container for physical storage
gold_location = "abfss://gold@stlinggarprojectdev001.dfs.core.windows.net/fact_trips_detail"
gold_table_name = "dev_gold.default.fact_trips_detail"

# 2. Read from the Silver Table
df_silver_ready = spark.read.table("dev_silver.default.silver_ridehailing")

# 3. Select and Rename for Business Reporting
# We use .alias() to make columns look professional for Power BI
df_gold = df_silver_ready.select(
    col("hvfhs_license_num").alias("License_Number"),
    col("dispatching_base_num").alias("Base_Number"),
    col("pickup_datetime").alias("Pickup_Time"),
    col("dropoff_datetime").alias("Dropoff_Time"),
    col("pulocationid").alias("Pickup_Zone_ID"),
    col("dolocationid").alias("Dropoff_Zone_ID"),
    col("sr_flag").alias("Shared_Ride_Flag"),
    col("ingestion_time").alias("Data_Ingestion_Timestamp"),
    col("silver_processed_time").alias("Data_Processed_Timestamp")
)

# 4. Write to Gold Table
(df_gold.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .option("path", gold_location) # Physical storage in Gold container
    .saveAsTable(gold_table_name)) # Registered name in Unity Catalog

print(f"Step 3 Complete: Gold Table '{gold_table_name}' is ready for Power BI!")
print(f"Final Row Count in Gold: {spark.read.table(gold_table_name).count()}")
```
